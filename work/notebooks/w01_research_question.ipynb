## 3. Quick look at the data (2-3 real numbers)

```python
import os
import sys
import subprocess

REPO_URL = "https://github.com/flyrank-bih/flyrank-ml-internship-starter"
REPO_DIR = "flyrank-ml-internship-starter"

if not os.path.isdir(REPO_DIR):
    subprocess.run(["git", "clone", "--depth", "1", REPO_URL, REPO_DIR], check=True)

os.chdir(REPO_DIR)

import pandas as pd
import numpy as np

df = pd.read_csv("data/raw/content_refresh_anonymized.csv")

df["is_declining_label"] = (
    df["trend_direction"].str.lower().eq("down").astype(int)
)

print("Dataset loaded:", df.shape)
```

```python
print("Total pages:", len(df))

print("Declining pages:",
      round(df["is_declining_label"].mean() * 100, 2), "%")

print("Pages with 180+ days since last update:",
      round((df["days_since_last_update"] >= 180).mean() * 100, 2), "%")
```

## 4. Careful words: what I can and can't claim

*Write what your work will be able to say (observed, directional, decision-support) — and what it never will (causal proof, 'predicting Google').*

- My analysis can identify pages that show observable signals associated with potential content-refresh opportunities, such as declining search performance, low CTR, high impressions, older content, or weak engagement. The results can be used as directional decision support to help prioritize which pages should be reviewed first. However, the analysis cannot prove that refreshing a page will cause better rankings, traffic, or engagement, and it cannot predict how Google will respond to a content change.

```python
# Compare declining vs non-declining pages using observable signals

summary = df.groupby("is_declining_label")[
    ["impressions_90d", "content_age_days", "avg_position", "ctr", "engagement_rate"]
].mean().round(2)

summary
```

### **What I can claim:**
- The data shows observable differences between declining and non-declining pages. For example, declining pages have lower average CTR (0.32 vs. 0.73) and lower impressions over 90 days (4,919 vs. 5,533). These signals can be used to prioritize pages for content review and provide directional decision support.

### **What I cannot claim:**
- These results show associations, not causation. I cannot claim that refreshing a page will cause its CTR or traffic to increase, or that the analysis can predict Google's future rankings.
